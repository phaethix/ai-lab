```python
import numpy as np
```

```python
A = np.array([[2, 1, -1], [-3, -1, 2], [-2, 1, 2]], dtype=float)
b = np.array([8, -11, -3], dtype=float)
```

```python
x = np.linalg.solve(A, b)
```

```python
eigenvalues, V = np.linalg.eig(A)
```

```python
D = np.diag(eigenvalues)
```
