## 线性代数：解线性方程组与特征分解

用 NumPy 求解 $A\mathbf{x} = \mathbf{b}$ 并做特征分解，验证 $A = VDV^{-1}$。

依赖：`pip install numpy`


```python
import numpy as np
```

### 定义方程组

$A$ 为 3×3 系数矩阵，$\mathbf{b}$ 为常向量；$A$ 非奇异，方程组存在唯一解。


```python
A = np.array([[2, 1, -1], [-3, -1, 2], [-2, 1, 2]], dtype=float)
b = np.array([8, -11, -3], dtype=float)
```

### 求解

`np.linalg.solve` 基于 LU 分解，比先求逆再相乘 `inv(A) @ b` 更快且数值更稳定。


```python
x = np.linalg.solve(A, b)
```

```python
x
```

### 特征分解

`np.linalg.eig` 返回特征值数组与特征向量矩阵 $V$（每列对应一个特征向量）。
其返回值恒为复数 dtype，因此即使特征值全为实数也会显示 `+0.j`。


```python
eigenvalues, V = np.linalg.eig(A)
```

### 构造对角矩阵

把特征值排成对角矩阵 $D$，与 $V$ 共同构成特征分解 $A = VDV^{-1}$。


```python
D = np.diag(eigenvalues)
```
